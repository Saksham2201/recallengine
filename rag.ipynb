```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_ollama import OllamaEmbeddings
```

```python
loader = PyPDFLoader(file_path="Documents/2608.08801v1.pdf",mode="page")
docs = loader.load()
docs[0].metadata["page_label"]
```

```python
lazy = loader.lazy_load()
docs_list = []
for doc in lazy:
    docs_list.append(doc)
    print(doc.metadata["page_label"])
```

```python
len(docs)
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=500,
    chunk_overlap=55,
    length_function=len)
texts = splitter.split_documents(docs)
```

```python
texts[20].page_content
```

```python
from termcolor import COLORS, colored
from random import choice
def display_chunks(chunks):
    colors_list = list(COLORS.keys())[2:8]
    print(f"Total Number of Chunks: {len(chunks)}")
    for chunk in chunks:
        # print(f"Chunk {num}: Length {len(chunk)} chars")
        print(colored(text=chunk.page_content, color=choice(colors_list)), end="\n\n")
```

```python
embeddings = OllamaEmbeddings(
    model="embeddinggemma:latest"
)
```

```python
question = "What is your name?"
single_vector = embeddings.embed_query(question)
```

```python
single_vector
```

```python
def preview_text(text, limit=120):
    """Return a short preview for cleaner notebook output."""
    if len(text) <= limit:
        return text
    return text[:limit] + "..."


def print_documents(title, docs):
    """Print retrieved documents using page metadata and a text preview."""
    print(title)
    for index, doc in enumerate(docs, start=1):
        print(f"{index}. page={doc.metadata.get('page')} | source={doc.metadata.get('source')}")
        print(f"   content={doc.page_content}")
    print()
```

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_documents(documents=texts,
    collection_name="rag_collection",
    embedding=embeddings,
    persist_directory="./chroma_langchain_db",
)
```

```python
query = "What is the title?"
results = vector_store.similarity_search(query, k=3)

print_documents("chunks",results)
```
